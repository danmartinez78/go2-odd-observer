# ODD Analysis - Production Workflow

This notebook demonstrates using the `odd_agents` shared module to run ODD (Operational Design Domain) analysis on robot sensor data.

**Key Features:**
- Single source of truth (imports from `odd_agents` module)
- Clean, minimal code
- Production-ready workflow

## Setup and Configuration

Import the ODD workflow from the shared module.

```python
import json
from odd_agents import run_odd_workflow

print("✅ odd_agents module loaded successfully")
```

## Run ODD Analysis

Execute the complete workflow on a scenario.

```python
# Run the workflow (uses default ODD description)
result = await run_odd_workflow(scenario_name="sim_run_test")

if result:
    print("✅ Workflow completed successfully!")
else:
    print("❌ Workflow failed")
```

## View Results

Display the executive summary and key findings.

```python
if result:
    report = result.get("report", {})
    
    print("=" * 80)
    print("EXECUTIVE SUMMARY")
    print("=" * 80)
    print(report.get("executive_summary", "N/A"))
    
    print("\n" + "=" * 80)
    print("KEY FINDINGS")
    print("=" * 80)
    for finding in report.get("key_findings", []):
        print(f"  • {finding}")
    
    print("\n" + "=" * 80)
    print("ODD COMPLIANCE")
    print("=" * 80)
    print(report.get("odd_compliance_summary", "N/A"))
```

## Full Report (JSON)

View the complete structured report.

```python
if result:
    print(json.dumps(result, indent=2))
```
